### Density Estimation

$$ \underbrace{\left\langle \log \frac{p(x)}{q_θ(x)} \right\rangle}_{I[p;q]} = \underbrace{\left\langle \log p(x) \right\rangle \vphantom{\left\langle \frac p q \right\rangle} }_{-H[p]} + \underbrace{\left\langle -\log q_θ(x) \right\rangle \vphantom{\left\langle \frac p q \right\rangle}}_{H[p;q]},  $$

so we can estimate how our model Q encodes the data from P. Since the first term is treated constant, we focus and minimize the second term which is the likelihood of data samples from $p$ under $q$.

In reality, $p$ is estimated and we need to work out issues to best estimate this objective.

### Variational Autoencoders

We wish to augment the real world with a random variable $Z$, our representation. We can parameterize this with $p(z | x)$. This is our encoder, which maps $X$ to a distribution for its representation. So $p(x, z) = p(z | x)p(x)$.

We seek a distribution if we can use the latents to formulate the images in the opposite direction. $q(x, z) = q(z)q(x | z)$. So we would draw a variable $z$ from a prior distribution and decode it using stochastic map $q(x | z)$ to generate images.

We try to minimize KL divergence between them. This gives us ELBO objective:
$$  \left\langle \log \frac{p(x,z)}{q(x,z)} \right\rangle_p = \left\langle \log \frac{p(x)p(z|x)}{q(x|z)q(z)} \right\rangle_p \geq 0. $$
